```python
# Program name: MOD2-Lab2-Lab.ipynb
# Author: Jenessa Rimer
# Date last updated: 11/05/2026
# Purpose: Get Student First and last name and gpa.
class InputVal:
    @staticmethod 
    def quitProcessing(LastNameinput) -> bool :
        if LastNameinput == "ZZZ":
            return False
        return True

    @staticmethod
    def Gpatest(gpa):
        if gpa >= 3.25:
            print("You have made the Honor Roll")
            if gpa >= 3.5 :
                print("You have made the Deans list")

lastname = input("Enter student Last Name: ")
while InputVal.quitProcessing(lastname):
    firstname = input("Enter student First Name: ")
    gpa = float(input("Enter student Gpa: "))
    print(f"Last Name: {lastname}\nFirst Name: {firstname}\nGPA: {gpa}")
    InputVal.Gpatest(gpa)
    lastname = input("Enter student Last Name: ")

print("PROGRAM ENDED")
```
